## Using Webscraping to Extract Stock Data Example


We will extract Netflix stock data [https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/netflix_data_webpage.html](https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/netflix_data_webpage.html).


Use the `requests` library to download the webpage [https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/amazon_data_webpage.html](https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/amazon_data_webpage.html). Save the text of the response as a variable named `html_data`.


```python
import pandas as pd  
import requests  
  
# Download the webpage using the requests library  
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0220EN-SkillsNetwork/labs/project/amazon_data_webpage.html"  
response = requests.get(url)  
  
# Save the text of the response as html_data  
html_data = response.text  
  
# Use pandas read_html to extract tables from the webpage  
read_html_pandas_data = pd.read_html(html_data)  
  
# Display the extracted tables  
for idx, table in enumerate(read_html_pandas_data):  
    print(f"Table {idx}:")  
    print(table.head())  
    print("\n")  
```

```python
import yfinance as yf  
  
# Extract stock data for Tesla  
tesla_data = yf.download('TSLA', start='2022-01-01', end='2022-12-31')  
  
# Reset the index  
tesla_data.reset_index(inplace=True)  
  
# Save the DataFrame to a CSV file  
tesla_data.to_csv('tesla_stock_data.csv', index=False)  
  
# Display the first five rows of the tesla_data DataFrame  
print(tesla_data.head())  
```

```python
# Display the last five rows of the tesla_data DataFrame  
print(tesla_data.tail())  
```

```python
# Extract stock data for GameStop (GME)  
gme_data = yf.download('GME', start='2022-01-01', end='2022-12-31')  
  
# Reset the index  
gme_data.reset_index(inplace=True)  
  
# Save the DataFrame to a CSV file  
gme_data.to_csv('gme_stock_data.csv', index=False)  
  
# Display the first five rows of the gme_data DataFrame  
print(gme_data.head())  
```

```python
# Display the last five rows of the gme_data DataFrame  
print(gme_data.tail())  
```

```python
import matplotlib.pyplot as plt  
  
def make_graph(data, title):  
    plt.figure(figsize=(12, 6))  
    plt.plot(data['Date'], data['Close'], label='Close Price')  
    plt.xlabel('Date')  
    plt.ylabel('Price')  
    plt.title(title)  
    plt.legend()  
    plt.grid(True)  
    plt.show()  
  
# Assuming tesla_data contains the stock data for Tesla  
make_graph(tesla_data, 'Tesla Stock Data')  
```
