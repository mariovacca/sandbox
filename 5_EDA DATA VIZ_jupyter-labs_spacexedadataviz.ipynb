### TASK  3: Visualize the relationship between success rate of each orbit type


Next, we want to visually check if there are any relationship between success rate and orbit type.


Let's create a `bar chart` for the sucess rate of each orbit


```python
import matplotlib.pyplot as plt

# Group by Orbit type and calculate the mean of Class column (success rate)
orbit_success_rate = df.groupby('Orbit')['Class'].mean()

# Create a bar chart
plt.figure(figsize=(10, 6))
orbit_success_rate.plot(kind='bar')

# Add labels and title
plt.xlabel('Orbit Type', fontsize=14)
plt.ylabel('Success Rate', fontsize=14)
plt.title('Success Rate by Orbit Type', fontsize=16)

# Improve readability
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

# Display the plot
plt.show()
```

Analyze the plotted bar chart to identify which orbits have the highest success rates.


### TASK  4: Visualize the relationship between FlightNumber and Orbit type

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create categorical plot with FlightNumber vs Orbit
sns.catplot(x="FlightNumber", y="Orbit", hue="Class", data=df, aspect=5)

# Add labels with increased font size
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Orbit Type", fontsize=20)

# Display the plot
plt.show()
```

### TASK  6: Visualize the launch success yearly trend


You can plot a line chart with x axis to be <code>Year</code> and y axis to be average success rate, to get the average launch success trend.


The function will help you get the year from the date:


```python
# A function to Extract years from the date 
year=[]
def Extract_year():
    for i in df["Date"]:
        year.append(i.split("-")[0])
    return year
Extract_year()
df['Date'] = year
df.head()
    
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Check if there's a date column in the dataframe
# Assuming there's a date column that needs year extraction
# If there's already a 'Year' column, we can skip this step
if 'Date' in df.columns:
    # Extract year from the Date column
    df['Year'] = pd.to_datetime(df['Date']).dt.year
elif 'LaunchDate' in df.columns:
    # If the date column is named LaunchDate
    df['Year'] = pd.to_datetime(df['LaunchDate']).dt.year

# Group by Year and calculate the mean of Class column (success rate)
yearly_success_rate = df.groupby('Year')['Class'].mean()

# Create a line chart
plt.figure(figsize=(12, 6))
yearly_success_rate.plot(kind='line', marker='o', linestyle='-')

# Add labels and title
plt.xlabel('Year', fontsize=14)
plt.ylabel('Success Rate', fontsize=14)
plt.title('Launch Success Rate by Year', fontsize=16)

# Improve readability
plt.grid(True, linestyle='--', alpha=0.7)
plt.ylim(0, 1.1)  # Set y-axis limit for better visualization of rates
plt.xticks(rotation=45)
plt.tight_layout()

# Display the plot
plt.show()
```

you can observe that the sucess rate since 2013 kept increasing till 2020

### TASK  7: Create dummy variables to categorical columns


Use the function <code>get_dummies</code> and <code>features</code> dataframe to apply OneHotEncoder to the column <code>Orbits</code>, <code>LaunchSite</code>, <code>LandingPad</code>, and <code>Serial</code>. Assign the value to the variable <code>features_one_hot</code>, display the results using the method head. Your result dataframe must include all features including the encoded ones.


```python
import pandas as pd

# Apply one-hot encoding to the categorical columns
# Note: Using pd.get_dummies() for one-hot encoding
features_one_hot = df.copy()
categorical_columns = ['Orbit', 'LaunchSite', 'LandingPad', 'Serial']

# Create dummy variables and add them to the dataframe
for column in categorical_columns:
    # Check if the column exists in the dataframe
    if column in df.columns:
        # Create dummies for this column
        dummies = pd.get_dummies(df[column], prefix=column, prefix_sep='_')
        # Add the dummy columns to our features_one_hot dataframe
        features_one_hot = pd.concat([features_one_hot, dummies], axis=1)

# Display the first few rows of the result
features_one_hot.head()
```

### TASK  8: Cast all numeric columns to `float64`


Now that our <code>features_one_hot</code> dataframe only contains numbers, cast the entire dataframe to variable type <code>float64</code>


```python
# Step 1: Identify which columns can be converted to float
numeric_columns = features_one_hot.select_dtypes(include=['int64', 'float64', 'int32', 'float32']).columns
categorical_columns = features_one_hot.select_dtypes(include=['object', 'category']).columns

print(f"Numeric columns: {len(numeric_columns)}")
print(f"Categorical columns that need handling: {len(categorical_columns)}")

if len(categorical_columns) > 0:
    print("Categorical columns found:")
    print(categorical_columns.tolist())

# Step 2: Handle remaining categorical columns
# Option 1: One-hot encode any remaining categorical columns
for column in categorical_columns:
    dummies = pd.get_dummies(features_one_hot[column], prefix=column, prefix_sep='_')
    features_one_hot = pd.concat([features_one_hot, dummies], axis=1)
    features_one_hot = features_one_hot.drop(column, axis=1)

# Step 3: Now convert all columns to float64
features_one_hot = features_one_hot.astype('float64')

# Display the first few rows and dtypes to verify conversion
print("\nData types after conversion:")
print(features_one_hot.dtypes)

# Show the first few rows of the dataframe
features_one_hot.head()
```
